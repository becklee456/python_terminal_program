```python
# logistic regression

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score


data = {
    "study_hours":[1,2,3,4,5,6],
    "attendance":[50,60,65,70,80,90],
    "pass":[0,0,0,1,1,1]
}

df=pd.DataFrame(data)

x=df[["study_hours","attendance"]]
y=df["pass"]

x_train,x_test,y_train,y_test = train_test_split (
    x,y,test_size=0.2,random_state=42
)

model=LogisticRegression()

model.fit(x_train,y_train)

y_pred=model.predict(x_test)


accuracy=accuracy_score(y_test,y_pred)
precision=precision_score(y_test,y_pred)
recall=recall_score(y_test,y_pred)
f1=f1_score(y_test,y_pred)


print("Accuracy",accuracy)
print("precision",precision)
print("recall",recall)
print("f1",f1)

new_student =pd.DataFrame( {
    "study_hours":[7],
    "attendance":[80]
})

new_predict=model.predict(new_student)

print(new_predict)

plt.figure()

# Pass students
plt.scatter(df[df["pass"]==1]["study_hours"],
            df[df["pass"]==1]["attendance"],
            label="Pass")

# Fail students
plt.scatter(df[df["pass"]==0]["study_hours"],
            df[df["pass"]==0]["attendance"],
            label="Fail")

plt.xlabel("Study Hours")
plt.ylabel("Attendance")
plt.title("Logistic Regression Data")
plt.legend()

plt.grid(True)
plt.show()
```
