```python
import plotly.express as pt
import plotly.graph_objects as go
data = {'months': ["Jan", "Feb", "March", "Apr"],
        'sales': [5000, 9000, 2000, 3000]}
grap = pt.scatter(data, x='months', y='sales', title="Monthly Sales")
grap.show()
```

```python
import matplotlib.pyplot as plt
Budget = [1000,3000,7000,10000,20000,40000]
Months = ["Jan","Feb","March","April","May","June"]
plt.barh(Months,Budget)
plt.title("barh plot")
plt.ylabel("Months")
plt.xlabel("Budget")
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt

Budget = [1000, 3000, 7000, 10000, 20000, 40000]
Months = ["Jan", "Feb", "March", "April", "May", "June"]

explode = [0, 0, 0, 0, 0.1, 0] 

plt.figure(figsize=(8, 8))
plt.pie(Budget, labels=Months, shadow=True, explode=explode ,autopct='%1.1f%%')
plt.title("Monthly Budget Pie Chart")

plt.show()
```

```python
import matplotlib.pyplot as plt
Budget = [1000,3000,7000,10000,20000,40000]
Months = ["Jan","Feb","March","April","May","June"]
plt.scatter(Budget,Months)
plt.title("scatter plot")
plt.ylabel("Months")
plt.xlabel("Budget")
plt.grid()
plt.show()
```

```python
import plotly.express as pt
data = {'months': ["Jan", "Feb", "March", "Apr"],
        'sales': [5000, 9000, 2000, 3000]} 

graph = pt.pie(data, names='months', values='sales',title="Monthly Sales" ,hole =0.5 )
graph.update_traces(pull=[0, 0.1, 0, 0.1])
graph.show()
```
